#### 1. Count Vowels and Consonants

##### . assume all characters are lowercase

##### . string may include special characters

```python
s = '__hel lo 123$world'
length_s = len(s.replace(' ','').replace('__','').replace("123",'').replace("$",''))

list1 = ['a','e','i','o','u']
count_vowels = 0
for char in s:
    if char in list1:
        count_vowels = count_vowels + 1
print(count_vowels)

count_consonants = length_s - count_vowels
print(count_consonants)

```

```python
vowels = 0
consonants = 0
for ch in s:
    if ch.isalpha():
        if ch in 'aeiou':
            vowels += 1
        else:
            consonants += 1

print(f" there are {vowels} vowels and {consonants} consonants")
```

#### 2. Count frequency of Characters

```python
s = "bananaaaabb"
```

##### dictionary

```python
# method 1
dict1 = {}
for ch in s:
    if ch in dict1.keys():
        dict1[ch] += 1
    else:
        dict1[ch] = 1

print(dict1)

# method 2
counts = dict()
for ch in s:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)
```

#### collections module

```python
from collections import Counter
Counter(s)
```

#### 3. Palindrome String

```python
s = "madamsss"
# method 1
print(s[::-1] == s[::1])

# method 2

def is_palindrome(text):
    left = 0
    right = len(text) - 1

    while left < right:
        if text[left] != text[right]:
            return "Not Palindrome"
        left = left + 1
        right = right - 1

    return "Palindrome"


is_palindrome(s)

```
