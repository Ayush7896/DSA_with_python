### Finding Maximum Sum

```python
def find_maximum_sum(arr):
    max_sum = 0
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            current_sum = arr[i] + arr[j]
            if current_sum > max_sum:
                max_sum = current_sum
    return max_sum


arr = [1,2,3,4,5]
result = find_maximum_sum(arr)
print("The maximum sum of any two elements is:", result)
```

### Find Maximum Element

```python
def find_maximum_element(arr):
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] > arr[j]:
                max_element = arr[i]
            else:
                max_element = arr[j]
    return max_element

arr = [1, 2, 3, 4, 5]
result = find_maximum_element(arr)
print("The maximum element in the array is:", result)
```

### Two Sum Using Two Pointer works for sorted Arrays

Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]
Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]
 

```python
def two_sum_two_pointer(array, target):
    left = 0
    right = len(array) - 1
    while left < right:
        curr_sum = array[left] + array[right]
        if curr_sum == target:
            return (left, right)
        elif curr_sum < target:
            left += 1
        else:
            right -= 1
    return None

array = [3,3]
target = 6
result = two_sum_two_pointer(array, target)
print(result)

# for unsorted arrays, it will fail, use hashmap
```

### Two Sum Using HashMap 

Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]
Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]
 

```python
def two_pointer_using_hashmap(array, target):
    seen = {}
    for i, number in enumerate(array):
        complement = target - number
        if complement in seen:
            return (seen[complement], i)
        seen[number] = i
        print(seen)
    return None 


array = [2,3,4,7]
target = 6
result = two_pointer_using_hashmap(array, target)
print(result)
```

### 3 Sum problem
#### Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.


Example 1:

Input: nums = [-1,0,1,2,-1,-4]

Output: [[-1,-1,2],[-1,0,1]]

Explanation: 

nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.

nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.

nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.

The distinct triplets are [-1,0,1] and [-1,-1,2].

Notice that the order of the output and the order of the triplets does not matter.


Example 2:

Input: nums = [0,1,1]

Output: []

Explanation: The only possible triplet does not sum up to 0.

Example 3:

Input: nums = [0,0,0]

Output: [[0,0,0]]

Explanation: The only possible triplet sums up to 0.
