```python
# 15. Running Average

# Calculate running average of list elements
# Example: [1,2,3,4,5] → [1, 1.5, 2, 2.5, 3]

list1 = [1,2,3,4,5]
sum1 = 0
result = []
for i in range(len(list1)):
    sum1 = sum1 + list1[i]
    avg = sum1/(i+1)
    result.append(avg)

print(result)
```

```python
# 16. Flatten List

# Flatten a shallow nested list
# Example: [[1,2],[3,4],[5,6]] → [1,2,3,4,5,6]
list1 = [[1,2],[3,4],[5,6]]
result = []
def flatten_list(lst):
    for item in lst:
        if isinstance(item, list):
            flatten_list(item)
        else:
            result.append(item)
    return result


answer = flatten_list(list1)
print(answer)
```

```python
# 17. Nested List Sum

# Sum all numbers in a nested list (arbitrary depth)
# Example: [1,[2,3],[4,[5,6]]] → 21
list1 = [1,[2,3],[4,[5,6]]]
result = []
def flatten_list(lst):
    for item in lst:
        if isinstance(item, list):
            flatten_list(item)
        else:
            result.append(item)
    return result


answer = flatten_list(list1)
print(sum(answer))
```

```python
# 21. Pack Consecutive Duplicates

# Pack consecutive duplicates into sublists
# Example: [1,1,2,3,3,3,4] → [[1,1],[2],[3,3,3],[4]]

list1 = [1,1,2,3,3,3,4]
result = []
current_group = [list1[0]]
for i in range(1,len(list1)):
    if list1[i] == list1[i-1]:
        current_group.append(list1[i])
    else:
        result.append(current_group)
        current_group = [list1[i]]

result.append(current_group)
print(result)
```

```python
# # Two sum
# Input: nums = [2, 7, 11, 15], target = 9
# Output: [0, 1]  // Because nums[0] + nums[1] = 2 + 7 = 9

# Visual:
# Index:  0   1   2   3
# Value: [2] [7] [11][15]
#         ↑   ↑
#         └───┘ These sum to 9!

list1 = [2,7,11, 15]
target =  9

def two_sum_two_pointer(array, target):

    left = 0
    right = len(array) - 1

    while left < right:
        curr_sum = array[left] + array[right]
        if curr_sum == target:
            return (left,right)

        elif curr_sum < target:
            left += 1
        else:
            right -= 1

    return None

two_sum_two_pointer(list1, target)


def two_sum_hashmap(array, target):

    dict1  = {}
    for index, value in enumerate(array):
        compliment = target - value
        if compliment in dict1:
            return(index, dict1[compliment])
        
        dict1[value] = index


two_sum_hashmap(list1, target)
```
