### Finding Maximum Sum

```python
def find_maximum_sum(arr):
    max_sum = 0
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            current_sum = arr[i] + arr[j]
            if current_sum > max_sum:
                max_sum = current_sum
    return max_sum


arr = [1,2,3,4,5]
result = find_maximum_sum(arr)
print("The maximum sum of any two elements is:", result)
```

### Find Maximum Element

```python
def find_maximum_element(arr):
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] > arr[j]:
                max_element = arr[i]
            else:
                max_element = arr[j]
    return max_element

arr = [1, 2, 3, 4, 5]
result = find_maximum_element(arr)
print("The maximum element in the array is:", result)
```

### Two Sum Using Two Pointer works for sorted Arrays

Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]
Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]
 

```python
def two_sum_two_pointer(array, target):
    left = 0
    right = len(array) - 1
    while left < right:
        curr_sum = array[left] + array[right]
        if curr_sum == target:
            return (left, right)
        elif curr_sum < target:
            left += 1
        else:
            right -= 1
    return None

array = [3,3]
target = 6
result = two_sum_two_pointer(array, target)
print(result)

# for unsorted arrays, it will fail, use hashmap
```

### Two Sum Using HashMap 

Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]
Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]
 

```python
def two_pointer_using_hashmap(array, target):
    seen = {}
    for i, number in enumerate(array):
        complement = target - number
        if complement in seen:
            return (seen[complement], i)
        seen[number] = i
        print(seen)
    return None 


array = [2,3,4,7]
target = 6
result = two_pointer_using_hashmap(array, target)
print(result)
```

```python
def sum(array, target):
    dict1 = {}
    for i, number in enumerate(array):
        compliment = target - number
        if compliment in dict1:
            return (dict1[compliment], i)
        dict1[number] = i
    return None

array = [2,3,4,7]
target = 6
result = sum(array, target)
print(result)
```

The distinct triplets are [-1,0,1] and [-1,-1,2].

Notice that the order of the output and the order of the triplets does not matter.

```python
def merge_intervals(intervals):
    if not intervals:                    # Line 1
        return []                        # Line 2
    
    intervals.sort(key=lambda x: x[0])   # Line 3 ← SORTING HAPPENS HERE
    merged = [intervals[0]]              # Line 4
    print(merged)
    
    for current in intervals[1:]:        # Line 5
        last_merged = merged[-1]         # Line 6
        
        if current[0] <= last_merged[1]: # Line 7
            merged[-1][1] = max(last_merged[1], current[1])  # Line 8
        else:
            merged.append(current)       # Line 9
    
    return merged                        # Line 10



intervals =  [[1,3],[2,6],[8,10],[15,18]]
result = merge_intervals(intervals)
print(result)
```

```python
def max_subarray_sum(nums, k):
    # Step 1: Calculate the sum of the first window
    window_sum = sum(nums[:k])
    max_sum = window_sum
    print(f"Initial window: {nums[:k]}, sum = {window_sum}")

    # Step 2: Slide the window from left to right
    for i in range(k, len(nums)):
        # remove the element that goes out (nums[i-k]) 
        # and add the new element coming in (nums[i])
        print(f"\nSliding window...")
        print(f"Before sliding: window_sum = {window_sum}")
        print(f"Remove nums[{i-k}] = {nums[i-k]}")
        print(f"Add nums[{i}] = {nums[i]}")

        window_sum += nums[i] - nums[i-k]  # update window sum
        print(f"After sliding: window_sum = {window_sum}")
        
        max_sum = max(max_sum, window_sum)
        print(f"Current max_sum = {max_sum}")

    return max_sum


nums = [2, 1, 5, 1, 3, 2]
k = 3
print("Final Answer =", max_subarray_sum(nums, k))
```

#### Find element with left side smaller/right side greater in an array


```python
l = [3,1,2,5,8,7,9]

for i in range(1,len(l)-1):
    flag = True

    for j in range(0,i):
        if l[j] > l[i]:
            flag = False
    
    for k in range(i+1, len(l)):
        if l[k] < l[i]:
            flag = False

    if flag:
        print(l[i])

# Time complexity on2
```
