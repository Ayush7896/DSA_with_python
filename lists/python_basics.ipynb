#### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
corrdinate1_x, corrdinate1_y = float(input("Your first number")),float(input("your second number"))
corrdinate2_x, corrdinate2_y = float(input("Your first number")),float(input("your second number"))

result = ((corrdinate2_x - corrdinate1_x)**2 + (corrdinate2_y - corrdinate1_y)**2)**0.5
print(result)
````

#### Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.

````python
principal, roi, time_period = float(input("Type your input")),float(input("Type your input")),float(input("Type your input"))
simple_interest = (principal * roi * time_period)/100
print(simple_interest)
````
